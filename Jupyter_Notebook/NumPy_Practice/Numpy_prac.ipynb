```python
def generate_synthetic_data(num_samples=1000, num_features=5):
    np.random.seed(42)
    X = np.random.randn(num_samples, num_features)
    true_weights = np.array([1.5, -2.0, 3.5, 0.5, -1.0])
    y = X @ true_weights + np.random.randn(num_samples) * 0.1
    return X, y

def train_linear_regression(X, y):
    X_b = np.c_[np.ones((X.shape[0], 1)), X]
    theta_best = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
    return theta_best

def predict(X, theta):
    X_b = np.c_[np.ones((X.shape[0], 1)), X]
    return X_b @ theta

if __name__ == "__main__":
    X_train, y_train = generate_synthetic_data()
    weights = train_linear_regression(X_train, y_train)
    
    X_test = np.array([
        [1.0, 0.5, -1.2, 0.3, 0.8],
        [-0.5, 1.2, 0.1, -0.4, 0.9]
    ])
    predictions = predict(X_test, weights)
    
    print("Calculated Weights:", weights)
    print("Predictions:", predictions)
```
